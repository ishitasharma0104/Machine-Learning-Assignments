```python
#Q2
prices= " 120, 45 , 300 ,80, 95 "
prices=[int(i.strip())for i in prices.split(",")]
total=sum(prices)
highest=max(prices)
average=total/len(prices)

print(f"Total: {total}\nhightest : {highest}\nAverage : {average:.2f}")
```
