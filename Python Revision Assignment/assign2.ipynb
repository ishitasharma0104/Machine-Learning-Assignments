```python
#Q2
prices= " 120, 45 , 300 ,80, 95 "
prices=[int(i.strip())for i in prices.split(",")]
total=sum(prices)
highest=max(prices)
average=total/len(prices)

print(f"Total: {total}\nhightest : {highest}\nAverage : {average:.2f}")
```

```python
# Q3
def check(username):
  username=username.strip().lower()
  if username.isalnum() and 4<len(username)<13 and username[0].isalpha():
    return "Valid username"
  return "Invalid username"
print(check(" Rahul_123 "))
```

```python
# Q4
sentence= "Riya went to the market to buy some apples"
words=sentence.split()
total=len(words)
count=0
for i in words:
  if i[-1].lower() in "aeiou":
    count+=1
longest=max(words,key=len)
print(f"Total : {total}\nCount : {count}\nLongest word : {longest}")
```

```python
# Q5
word="Aba c"
freq={}
for i in word.lower():
  if i!=" ":
    if i in freq:
        freq[i]+=1
    else:
        freq[i]=1
print(freq)
```
