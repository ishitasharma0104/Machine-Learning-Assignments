```python
# Q1
def valide(input):
  input=input.strip()
  if len(input)>=5 and input.isalnum():
     return input.lower()
  else:
     return "Invalid"
print(valide(" CodeMaster99 "))
print(valide(" Python_Dev "))

```

```python
# Q2
def secCode(code):
  code = code.split("-")
  code = code[::-1]
  code = " ".join(code)
  code = code.upper()

  return code

print(secCode("charlie-bravo-alpha"))
```

```python
# Q3
def cleaner(data):
  data=data.split(";")
  data=[name.strip().title() for name in data]
  data=", ".join(data)
  return data
print(cleaner(" john doe ; jAnE sMiTh;ALAN tUrI"))
```

```python
# Q4
def eliminator(text):
  text=text.strip()
  text=text.split()
  text=" ".join(text)
  text=text.capitalize()
  

  return text
print(eliminator( " python is a very powerful language "))
```
