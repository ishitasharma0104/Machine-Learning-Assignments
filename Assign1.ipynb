```python
# Q1
def valide(input):
  input=input.strip()
  if len(input)>=5 and input.isalnum():
     return input.lower()
  else:
     return "Invalid"
print(valide(" CodeMaster99 "))
print(valide(" Python_Dev "))

```

```python
# Q2
def secCode(code):
  code = code.split("-")
  code = code[::-1]
  code = " ".join(code)
  code = code.upper()

  return code

print(secCode("charlie-bravo-alpha"))
```

```python
# Q3
def cleaner(data):
  data=data.split(";")
  data=[name.strip().title() for name in data]
  data=", ".join(data)
  return data
print(cleaner(" john doe ; jAnE sMiTh;ALAN tUrI"))
```

```python
# Q4
def eliminator(text):
  text=text.strip()
  text=text.split()
  text=" ".join(text)
  text=text.capitalize()
  

  return text
print(eliminator( " python is a very powerful language "))
```

```python
# Q5
def check(a):
  s=""
  for i in a:
    if i.isalnum():
      s+=i
  s=s.lower()
  rev=s[::-1]
  if s==rev:
    return True
  else:
    return False
print(check("Was it a car or a cat I saw?"))
print(check("Python is fun"))
```

```python
# Q6
def social(post):
  p=post.split()
  hash=[]
  for i in p:
    if i.startswith("#"):
      hash.append(i[1:].lower())
  return hash
print(social("Loving the SUNSET! #Nature #Sunset_Vi #beautiful"))


```

```python
# Q7
def extension(file):
  file=file.split(",")
  ext=[]
  for i in file:
    i=i.strip()
    i=i.split(".")[-1].upper()
    ext.append(i)
  ext=set(ext)
  return list(ext)
print(extension("report.pdf, image.JPG, data.csv, notes.txt, photo"))
  
    
```

```python
# Q8
def slicer(email):
  if "@" not in email:
    return "Invalid"
  
  email=email.split("@")[1].split(".")[0].title()
  return email
  

print(slicer("developer_99@microsoft.com" ))
print(slicer("admin_at_website.org"))
```
